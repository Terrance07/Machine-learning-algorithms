1. Output of a neuron using one input, weight, and bias

```python
input = 5
weight = 0.5
bias = 1
Output = (input * weight) + bias
print(Output)
```

2. Weighted sum for multiple inputs, weights, and bias

```python
x1 = 10
x2 = 15
w1 = 0.5
w2 = 0.6
b = 1
z = (x1 * w1) + (x2 * w2) + b
print(z)
```

3. ReLU activation function

```python
def relu(x):
    return max(0, x)

print(relu(5))
print(relu(-3))
```

4. Sigmoid activation function

```python
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

print(sigmoid(2))
print(sigmoid(-2))
```

6. Simple neuron to predict whether a student will pass

```python
study_hours = 6
attendance = 80
assignment_score = 70
weight1 = 0.4
weight2 = 0.3
weight3 = 0.3
bias = -50
weighted_sum = (study_hours * weight1) + \
               (attendance * weight2) + \
               (assignment_score * weight3) + bias
if weighted_sum >= 0:
    print("Pass")
else:
    print("Fail")
```

7. Calculate Mean Squared Error (MSE)

```python
actual = [10, 20, 30]
predicted = [12, 18, 29]
total = 0
for i in range(len(actual)):
    error = actual[i] - predicted[i]
    total += error ** 2
mse = total / len(actual)
print("MSE:", mse)
```

8. Compare loss of two predictions

```python
actual = 10
prediction1 = 8
prediction2 = 9
loss1 = (actual - prediction1) ** 2
loss2 = (actual - prediction2) ** 2
print("Loss 1:", loss1)
print("Loss 2:", loss2)
if loss1 < loss2:
    print("Prediction 1 is better")
else:
    print("Prediction 2 is better")
```
