```python
import pandas as pd
from sklearn.metrics import mean_squared_error
```

```python
actual = [10, 20, 30]
predicted = [12, 18, 33]
```

```python
mse = mean_squared_error(actual, predicted)
```

```python
print("Mean Squared Error:", mse)
```
