```python
import pandas as pd
from sklearn.metrics import mean_absolute_error
```

```python
actual = [14, 20, 30]
predicted = [12, 18, 33]
```

```python
mae = mean_absolute_error(actual, predicted)
```

```python
print("Mean Absolute Error:", mae)
```
